```python
import pandas as pd
```

```python
file_path = "../data/all_TEreservations_clean.csv"  
df = pd.read_csv(file_path, sep=';')
classgroups_series = df['Classgroups'].dropna().apply(lambda x: x.split('|'))

unique_classgroups = set([item for sublist in classgroups_series for item in sublist])

unique_classgroups_df = pd.DataFrame(unique_classgroups, columns=['Classgroup'])

output_file = "../data/unique_classgroups.csv"
unique_classgroups_df.to_csv(output_file, index=False,header=False)

print(f"{len(unique_classgroups)} classgroups from schedule to {output_file}")
```
