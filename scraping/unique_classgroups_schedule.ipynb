```python
import pandas as pd
import os

file_path = "../data/AllLectures.csv"
df = pd.read_csv(file_path, sep=';')

# Split and flatten classgroups
classgroups_series = df['Classgroups'].dropna().apply(lambda x: x.split('|'))
unique_classgroups = sorted(set(item.strip() for sublist in classgroups_series for item in sublist))

unique_classgroups_df = pd.DataFrame(unique_classgroups, columns=['Classgroup'])

output_file = "../data/unique_classgroups.csv"

# If file exists, read existing entries to avoid duplicates
if os.path.exists(output_file):
    existing = pd.read_csv(output_file, header=None)[0].tolist()
    unique_classgroups = sorted(set(unique_classgroups) - set(existing))
    unique_classgroups_df = pd.DataFrame(unique_classgroups, columns=['Classgroup'])

# Write only truly new unique classgroups
if not unique_classgroups_df.empty:
    unique_classgroups_df.to_csv(output_file, index=False, header=False, mode='a')

print(f"{len(unique_classgroups)} new unique classgroups written to {output_file}")
```
