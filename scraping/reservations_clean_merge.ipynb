```python
import os
import argparse
import pandas as pd
import shutil
from datetime import datetime
from dateutil import parser as dateparser
import glob
```

```python
# Archive helper: call this AFTER processing a file successfully.
ARCHIVE_DIR = "../data/archived"   
os.makedirs(ARCHIVE_DIR, exist_ok=True)

def _unique_target_path(target_dir, filename):
    dest = os.path.join(target_dir, filename)
    if not os.path.exists(dest):
        return dest
    ts = datetime.now().strftime("%Y%m%dT%H%M%SZ")
    name, ext = os.path.splitext(filename)
    return os.path.join(target_dir, f"{name}__{ts}{ext}")

def archive_file(src_path, archive_dir=ARCHIVE_DIR):
    if not os.path.exists(src_path):
        raise FileNotFoundError(f"Source not found: {src_path}")
    os.makedirs(archive_dir, exist_ok=True)
    filename = os.path.basename(src_path)
    dest = _unique_target_path(archive_dir, filename)
    try:
        shutil.move(src_path, dest)
        return dest
    except Exception as e:
        # fallback copy+remove
        try:
            shutil.copy2(src_path, dest)
            os.remove(src_path)
            return dest
        except Exception as e2:
            raise RuntimeError(f"Failed to archive {src_path}. move error: {e}; copy fallback: {e2}")
```

```python
DATA_FOLDER = "../data"
INPUT_FOLDER = "../data/reservations"
os.makedirs(DATA_FOLDER, exist_ok=True)
OUTPUT_FILE = os.path.join(DATA_FOLDER, "all_TEreservations_clean.csv")
```

```python
csv_files = sorted(glob.glob(os.path.join(INPUT_FOLDER, "TEreservations_*.csv")))

if not csv_files:
    print("No matching CSV files found in folder 'data'.")
else:
    print(f"Found {len(csv_files)} CSV files:")
    for f in csv_files:
        print("  -", os.path.basename(f))
```

```python
if csv_files:
    sample_file = csv_files[0]
    print(f"Inspecting sample file: {os.path.basename(sample_file)}")
    
    df_sample = pd.read_csv(sample_file, sep=';', low_memory=False)
    display(df_sample.head(10))
else:
    print("No sample file to inspect.")
```

```python
all_data = []
for file in csv_files:
    try:
        temp_df = pd.read_csv(file, sep=';', low_memory=False)
        temp_df['source_file'] = os.path.basename(file)
        all_data.append(temp_df)
        try:
            archived_path = archive_file(file)
            print(f"Archived processed file: {os.path.basename(file)} -> {archived_path}")
        except Exception as ar_err:
            print(f"Warning: failed to archive {file}: {ar_err}")

    except Exception as e:
        print(f"Error reading {file}: {e}")

if all_data:
    df = pd.concat(all_data, ignore_index=True)
    print(f"Combined {len(all_data)} files — total rows: {len(df):,}")
else:
    df = pd.DataFrame()
    print("No data combined.")
```

```python
if not df.empty:
    relevant_cols = [
        'Id', 'Start', 'End', 'Updated', 'UpdatedBy', 'Status',
        'WorkFormCourse', 'WorkFormExam', 'WorkFormEvaluation',
        'LearningEnvironment', 'ExamEnvironment', 'Classgroups',
        'TeacherIds', 'OlodPointers', 'Rooms', 'TrajectoryPart'
    ]
    
    # Keep only relevant columns and existing ones
    df_clean = df[[c for c in relevant_cols if c in df.columns]].copy()

    # Remove cancelled or missing status rows
    df_clean = df_clean[df_clean['Status'].str.lower() != 'cancelled']

    # Convert date columns to datetime if possible
    for col in ['Start', 'End', 'Updated']:
        if col in df_clean.columns:
            df_clean[col] = pd.to_datetime(df_clean[col], errors='coerce', dayfirst=True)

    print(f"Cleaned data shape: {df_clean.shape}")
    display(df_clean.head(10))
else:
    print("No data to clean.")
```

```python
if not df_clean.empty:
    df_clean.to_csv(OUTPUT_FILE, index=False, sep=';')
    print(f"Saved cleaned file to: {OUTPUT_FILE}")
else:
    print("Nothing to save — cleaned DataFrame is empty.")
```

```python
if not df_clean.empty:
    print("Summary:")
    print(f"Total rows: {len(df_clean):,}")
    print(f"Unique Teachers: {df_clean['UpdatedBy'].nunique() if 'UpdatedBy' in df_clean else 0}")
    print(f"Date range: {df_clean['Start'].min()} → {df_clean['End'].max()}")
    print("\nMissing values per column:")
    display(df_clean.isna().sum())
else:
    print("No data to summarize.")
```
