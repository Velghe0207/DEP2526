```python
import pandas as pd

df = pd.read_csv("../data/all_students.csv")

# Select only the columns we need
# Select distinct subgroups - reset index
subgroepid = df[['SUBGROEPID', 'SUBGROEPCODE']].drop_duplicates().reset_index(drop=True)



subgroepid.to_csv("../data/cleaned/subgroups/unique_subgroeps.csv", index=False)
```
