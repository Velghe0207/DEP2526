```python
import pandas as pd

df = pd.read_excel("../data/FMIS_lokalen.xlsx",header=1)
df
```

```python
gschb_df = df[df['Gebouw-beschrijving'].str.contains("GSCHB", na=False)]
gschb_df
```

```python
gschb_df["Categorie"].value_counts()
```

```python
# Unieke GSCHB lokalen opslaan in csv bestand
unique_rooms = gschb_df["Lokaal-Beschrijving"].unique()
unique_df = pd.DataFrame(unique_rooms, columns=["Lokaal-Beschrijving"])
unique_df
```

```python
unique_df.to_csv("../data/unique_gschb_rooms.csv", index=False)
```
